# 结果

```python
data = pd.read_excel(r'*****')
```

```python
data['LLM-结局'] = data['结果-值']>0.5
data['LLM-结局'] = data['LLM-结局'].replace(True,1).replace(False,0) 
```

```python
y_true = data['结局']
y_pred = data['LLM-结局']
y_prob = data['结果-值']

# 计算AUC
auc = roc_auc_score(y_true, y_prob)

# 计算精确率
pre = precision_score(y_true, y_pred)

# 计算F1分数
f1 = f1_score(y_true, y_pred)

# 计算准确率
acc = accuracy_score(y_true, y_pred)

# 计算灵敏度（召回率）
sen = recall_score(y_true, y_pred)

# 计算特异性
tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()
spe = tn / (tn + fp)

# 计算混淆矩阵
conf_matrix = confusion_matrix(y_true, y_pred)

# 打印结果
print(f"AUC: {auc}")
print(f"Precision (PRE): {pre}")
print(f"F1 Score (F1): {f1}")
print(f"Accuracy (ACC): {acc}")
print(f"Sensitivity (SEN): {sen}")
print(f"Specificity (SPE): {spe}")
print(f"Confusion Matrix:\n{conf_matrix}")
```
